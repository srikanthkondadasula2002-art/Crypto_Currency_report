```python
import pandas as pd
import matplotlib.pyplot as plt
from datetime import datetime

print("Task 1 notebook started successfully.")
```

```python
data = {
    "coin": [
        "Bitcoin", "Ethereum", "BNB", "Cardano", "Dogecoin",
        "Avalanche", "Bitcoin Cash", "Chainlink", "Cosmos", "Dai",
        "Polkadot", "Solana", "Uniswap", "USD Coin", "XRP",
        "Litecoin", "Polygon", "Tether", "Binance USD", "Decentraland"
    ],
    "symbol": [
        "BTC", "ETH", "BNB", "ADA", "DOGE",
        "AVAX", "BCH", "LINK", "ATOM", "DAI",
        "DOT", "SOL", "UNI", "USDC", "XRP",
        "LTC", "MATIC", "USDT", "BUSD", "MANA"
    ],
    "price": [
        65000, 3500, 600, 0.45, 0.15,
        35, 400, 15, 8, 1,
        7, 150, 10, 1, 0.55,
        80, 0.75, 1, 1, 0.4
    ],
    "24h_volume": [
        32000000000, 18000000000, 3500000000, 900000000,
        2500000000, 1200000000, 800000000, 1100000000,
        500000000, 400000000, 700000000, 4500000000,
        650000000, 3000000000, 2200000000, 600000000,
        1000000000, 28000000000, 500000000, 350000000
    ],
    "mkt_cap": [
        1200000000000, 420000000000, 90000000000,
        16000000000, 22000000000, 14000000000,
        8000000000, 9000000000, 3000000000, 5000000000,
        10000000000, 70000000000, 6000000000, 30000000000,
        30000000000, 6000000000, 7000000000, 110000000000,
        8000000000, 700000000
    ]
}

df = pd.DataFrame(data)

print("Dataset created successfully.")
print("Number of rows:", len(df))

display(df.head())
```

```python
allowed_letters = set("AEIOUBCD")

df["starting_letter"] = (
    df["coin"]
    .str.strip()
    .str[0]
    .str.upper()
)

eligible_coins = df[
    df["starting_letter"].isin(allowed_letters)
].copy()

print("Eligible coins:", len(eligible_coins))

display(
    eligible_coins[
        ["coin", "symbol", "price", "24h_volume"]
    ]
)
```

```python
top10 = (
    eligible_coins
    .sort_values(
        by="24h_volume",
        ascending=False
    )
    .head(10)
    .reset_index(drop=True)
)

top10.insert(
    0,
    "Rank",
    range(1, len(top10) + 1)
)

display(
    top10[
        ["Rank", "coin", "symbol", "price", "24h_volume"]
    ]
)
```

```python
plt.figure(figsize=(12, 7))

plt.barh(
    top10["coin"],
    top10["24h_volume"]
)

plt.xlabel("24h Volume")
plt.ylabel("Crypto Coin")

plt.title(
    "Top 10 Crypto Coins by 24h Volume"
)

plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()
```

```python
excel_file = "Crypto_Task1_Data.xlsx"

with pd.ExcelWriter(
    excel_file,
    engine="openpyxl"
) as writer:

    df.to_excel(
        writer,
        sheet_name="Data",
        index=False
    )

    eligible_coins.to_excel(
        writer,
        sheet_name="Analysis",
        index=False
    )

    top10.to_excel(
        writer,
        sheet_name="Top10",
        index=False
    )

print("Excel file created successfully:")
print(excel_file)
```

```python
import pandas as pd

crypto = pd.read_excel(
    "Crypto_Task1_ReadOnly.xlsx",
    sheet_name="Data",
    engine="openpyxl"
)

print("Data loaded successfully.")
print("Rows:", len(crypto))

display(crypto.head())
```

```python
valid_starts = tuple("AEIOUBCD")

selected = crypto[
    crypto["coin"]
    .astype(str)
    .str.strip()
    .str.upper()
    .str.startswith(valid_starts)
].copy()

print("Eligible coins:", len(selected))

display(
    selected[
        ["coin", "symbol", "price", "24h_volume"]
    ].head(20)
)
```

```python
# Rank the eligible coins by 24-hour trading volume
top10 = (
    selected
    .sort_values("24h_volume", ascending=False)
    .head(10)
    .reset_index(drop=True)
)

top10.insert(0, "Rank", range(1, len(top10) + 1))

print("Top 10 crypto coins by 24h volume:")

display(
    top10[
        ["Rank", "coin", "symbol", "price", "24h_volume"]
    ]
)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 7))

plt.barh(
    top10["coin"],
    top10["24h_volume"]
)

plt.xlabel("24h Trading Volume")
plt.ylabel("Crypto Coin")
plt.title("Top 10 Crypto Coins by 24h Volume")

plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()
```

```python
# Save the Task 1 Top 10 results

output_file = "Task1_Top10_Crypto.xlsx"

top10[
    ["Rank", "coin", "symbol", "price", "24h_volume"]
].to_excel(
    output_file,
    index=False
)

print("Excel file created successfully:")
print(output_file)
```

```python
from openpyxl import load_workbook
from openpyxl.chart import BarChart, Reference

output_file = "Task1_Top10_Crypto.xlsx"

wb = load_workbook(output_file)
ws = wb["Sheet1"]

# Create Dashboard
dashboard = wb.create_sheet("Dashboard")

dashboard["A1"] = "Crypto Time-Based Visibility & Sentiment Analysis"
dashboard["A3"] = "Please open in working hours (9 am to 5 pm)."

# Create chart
chart = BarChart()
chart.type = "bar"
chart.title = "Top 10 Crypto Coins by 24h Volume"
chart.x_axis.title = "24h Trading Volume"
chart.y_axis.title = "Crypto Coin"

# Volume = column E, coin names = column B
values = Reference(ws, min_col=5, min_row=1, max_row=11)
categories = Reference(ws, min_col=2, min_row=2, max_row=11)

chart.add_data(values, titles_from_data=True)
chart.set_categories(categories)

chart.height = 8
chart.width = 16

dashboard.add_chart(chart, "A6")

wb.save(output_file)

print("Dashboard created successfully.")
```
